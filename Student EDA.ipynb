```python
df.drop_duplicates(inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["G3"], bins=20, kde=True)
plt.title("Distribution of Final Grades")
plt.show()
```

```python
sns.boxplot(x="sex", y="G3", data=df)
plt.title("Gender vs Final Grade")
plt.show()
```

```python
sns.boxplot(x="G1", y="G3", data=df)
plt.title("Study Time Impact")
plt.show()
```

```python
sns.scatterplot(x="weekend_alcohol_consum", y="G3", data=df)
plt.title("Alcohol vs Grades")
plt.show()
```

```python
plt.figure(figsize=(12,8))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
corr["G3"].sort_values(ascending=False)
```

```python
sns.pairplot(df[["G1", "G2", "G3"]])
plt.show()
```
